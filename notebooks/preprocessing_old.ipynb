```python
# How many missing parts per row?
miss_k = comp.isna().sum(axis=1)
miss_k.value_counts()
```

```python
# Case A: exactly 1 missing -> fill with 100 - sum(others)
mask_one = (miss_k == 1)
sum_obs = comp.sum(axis=1, skipna=True)
fill_val = 100 - sum_obs

for col in comp_cols:
    m = mask_one & comp[col].isna()
    comp.loc[m, col] = fill_val[m]
```

```python
# Case B: no missing but sum != 100 -> rescale to sum to 100
row_sum = comp.sum(axis=1)
mask_full = (miss_k == 0) & row_sum.notna() & (row_sum > 0)

comp.loc[mask_full, comp_cols] = (
    comp.loc[mask_full, comp_cols]
    .div(row_sum[mask_full], axis=0) * 100
)

# Safety: keep within [0, 100]
comp[comp_cols] = comp[comp_cols].clip(0, 100)

# Write back
df_sch_sea2022_common[comp_cols] = comp
```

```python
def apply_schema_groups(df: pd.DataFrame, schema_groups):
    df = df.copy()
    for g in schema_groups:
        cols = [c for c in g["cols"] if c in df.columns]
        if not cols:
            continue

        # ensure numeric
        df[cols] = df[cols].apply(pd.to_numeric, errors="coerce")

        # apply post rules
        post = g.get("post", [])
        if "round" in post:
            df[cols] = df[cols].round()

        if "clip" in post:
            df[cols] = df[cols].clip(lower=g.get("min", None), upper=g.get("max", None))

        # enforce integer type for discrete types (optional but recommended)
        if g.get("type") in {"ordinal", "binary", "count"}:
            df[cols] = df[cols].astype("float")  # keeps NaN-safe integer dtype

    return df

df_sch_sea2022_common = apply_schema_groups(df_sch_sea2022_common, SCHEMA_GROUPS)
```

```python
comp = df_sch_sea2022_common[comp_cols].copy()
miss_k = comp.isna().sum(axis=1)

mask_train = (miss_k == 0)
mask_hard = (miss_k >= 2)

print("Rows needing ILR:", int(mask_hard.sum()))
```

```python
# For ILR based imputation helpers

def closure(X, total=100.0):
    X = np.asarray(X, dtype=float)
    row_sums = X.sum(axis=1, keepdims=True)
    return (X / row_sums) * total

def make_strictly_positive(X, eps=1e-6):
    X = np.asarray(X, dtype=float)
    return np.where(X <= 0, eps, X)

def clr(X):
    logX = np.log(X)
    gm = logX.mean(axis=1, keepdims=True)
    return logX - gm

def helmert_submatrix(D):
    H = np.zeros((D - 1, D))
    for i in range(1, D):
        H[i - 1, :i] = 1 / np.sqrt(i * (i + 1))
        H[i - 1, i] = -i / np.sqrt(i * (i + 1))
    return H

def ilr_transform(X):
    D = X.shape[1]
    H = helmert_submatrix(D)
    return clr(X) @ H.T

def ilr_inverse(Z):
    D = Z.shape[1] + 1
    H = helmert_submatrix(D)
    clrX = Z @ H
    return np.exp(clrX)
```

```python
# Training compositions
comp_train = comp.loc[mask_train, comp_cols].to_numpy()
comp_train = make_strictly_positive(comp_train)
comp_train = closure(comp_train)

Z_train = ilr_transform(comp_train)

# Predictors (non_comp)
predictors_train = df_sch_sea2022_common.loc[mask_train, non_comp_cols]
predictors_test = df_sch_sea2022_common.loc[mask_hard, non_comp_cols]

Z_cols = [f"ILR_{i}" for i in range(Z_train.shape[1])]

train_df = predictors_train.copy()
for i, col in enumerate(Z_cols):
    train_df[col] = Z_train[:, i]

test_df = predictors_test.copy()
for col in Z_cols:
    test_df[col] = np.nan

combo = pd.concat([train_df, test_df], ignore_index=True)
```

```python
it = IterativeImputer(
    estimator=ExtraTreesRegressor(n_estimators=200, random_state=42),
    max_iter=10,
    random_state=42
)

combo_imp = it.fit_transform(combo)
combo_imp = pd.DataFrame(combo_imp, columns=combo.columns)

Z_imputed = combo_imp.loc[len(train_df):, Z_cols].to_numpy()
```

```python
comp_back = ilr_inverse(Z_imputed)
comp_back = make_strictly_positive(comp_back)
comp_back = closure(comp_back)

comp.loc[mask_hard, comp_cols] = comp_back
df_sch_sea2022_common[comp_cols] = comp
```

```python
def select_student_columns(df):
    """
    Select base student columns + plausible values (PV) for READ, SCIE, MATH.
    """
    
    # Automatically detect PV columns
    pv_cols = [
        col for col in df.columns
        if col.startswith("PV") and (
            col.endswith("READ") or 
            col.endswith("SCIE") or 
            col.endswith("MATH")
        )
    ]
    
    # Combine
    selected_cols = stu_cols_to_use + pv_cols
    
    # Keep only those that exist
    selected_cols = [col for col in selected_cols if col in df.columns]
    
    return df[selected_cols]
```

```python
COUNTRY_COL = "CNT"
SCHOOL_COL  = "CNTSCHID"
WEIGHT_COL  = "W_FSTUWT"
DOMAINS = ["MATH", "READ", "SCIE"]

def get_pv_cols(df, domain):
    pv = [c for c in df.columns if c.startswith("PV") and c.endswith(domain)]
    # sort PV1..PV10 by the number
    return sorted(pv, key=lambda x: int(x.replace("PV", "").replace(domain, "")))

def pisa_school_weighted_by_domain(stu_df):
    """
    Returns one row per (CNT, CNTSCHID) with:
      - <DOMAIN>_school_weighted (PISA-style: compute per PV then average over PVs)
      - optional per-PV school means can be kept if you want
    """
    df = stu_df.copy()
    group_cols = [COUNTRY_COL, SCHOOL_COL]

    # base output skeleton (unique schools)
    out = df[group_cols].drop_duplicates().copy()

    # ensure weight is numeric
    df[WEIGHT_COL] = pd.to_numeric(df[WEIGHT_COL], errors="coerce")

    for domain in DOMAINS:
        pv_cols = get_pv_cols(df, domain)
        if not pv_cols:
            continue

        per_pv_school = []

        for pv in pv_cols:
            x = pd.to_numeric(df[pv], errors="coerce")
            w = df[WEIGHT_COL]

            tmp = pd.DataFrame({
                COUNTRY_COL: df[COUNTRY_COL],
                SCHOOL_COL:  df[SCHOOL_COL],
                "xw": x * w,
                "w":  w
            }).dropna()

            # school weighted mean for this PV
            school_pv = (
                tmp.groupby(group_cols, as_index=False)
                   .agg(xw_sum=("xw", "sum"), w_sum=("w", "sum"))
            )
            school_pv[pv + "_school"] = school_pv["xw_sum"] / school_pv["w_sum"]
            school_pv = school_pv[group_cols + [pv + "_school"]]

            per_pv_school.append(school_pv)

        # merge all PV-school means for the domain
        dom_df = per_pv_school[0]
        for d in per_pv_school[1:]:
            dom_df = dom_df.merge(d, on=group_cols, how="outer")

        # PISA-style combine: mean across PV-school means
        pv_school_cols = [c for c in dom_df.columns if c.endswith("_school") and c.startswith("PV")]
        dom_df[f"PV{domain}_weighted"] = dom_df[pv_school_cols].mean(axis=1)

        # keep only final domain score (or keep pv_school_cols too if you want)
        dom_df = dom_df[group_cols + [f"PV{domain}_weighted"]]

        out = out.merge(dom_df, on=group_cols, how="left")

    # overall across domains (optional)
    dom_cols = [c for c in out.columns if c.endswith("_weighted")]
    if dom_cols:
        out["ALL_DOMAINS_weighted"] = out[dom_cols].mean(axis=1)

    return out
```

## 2018 recomputed

```python
COUNTRY_COL = "CNT"
WEIGHT_COL  = "W_FSTUWT"
DOMAINS     = ["MATH", "READ", "SCIE"]

def get_pv_cols(df, domain):
    pv = [c for c in df.columns if c.startswith("PV") and c.endswith(domain)]
    return sorted(pv, key=lambda x: int(x.replace("PV", "").replace(domain, "")))

def weighted_mean(values, weights):
    return (values * weights).sum() / weights.sum()

def pisa_country_means(stu_df):
    """
    Returns one row per CNT with T_bar (pooled PV mean) per domain.
    SE computation excluded — validate means against OECD report first.
    """
    df = stu_df.copy()
    df[WEIGHT_COL] = pd.to_numeric(df[WEIGHT_COL], errors="coerce")

    results = []

    for cnt, grp in df.groupby(COUNTRY_COL):
        row = {COUNTRY_COL: cnt}

        for domain in DOMAINS:
            pv_cols = get_pv_cols(grp, domain)
            if not pv_cols:
                continue

            w = grp[WEIGHT_COL]
            T_d_list = []

            # Step 1: one weighted mean per PV
            for pv in pv_cols:
                x    = pd.to_numeric(grp[pv], errors="coerce")
                mask = x.notna() & w.notna()
                T_d_list.append(weighted_mean(x[mask], w[mask]))

            # Step 2: pool → final country mean (T_bar)
            row[f"{domain}_mean"] = sum(T_d_list) / len(T_d_list)

        results.append(row)

    return pd.DataFrame(results)
```
